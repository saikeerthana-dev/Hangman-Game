```python
import random

words = ["python", "computer", "programming", "hangman", "coding"]

word = random.choice(words)
guessed = []
wrong = 6

print("🎮 Welcome to Hangman Game!")

while wrong > 0:
    display = ""

    for letter in word:
        if letter in guessed:
            display += letter + " "
        else:
            display += "_ "

    print("\nWord:", display)

    if all(letter in guessed for letter in word):
        print("🎉 You won! The word was:", word)
        break

    guess = input("Enter a letter: ").lower()

    if guess in guessed:
        print("You already guessed that letter!")
        continue

    guessed.append(guess)

    if guess in word:
        print("✅ Correct guess!")
    else:
        wrong -= 1
        print("❌ Wrong guess!")
        print("Chances left:", wrong)

else:
    print("\n😢 Game Over!")
    print("The word was:", word)
```
