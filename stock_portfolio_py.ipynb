```python

portfolio = []
total = 0

n = int(input("Enter number of stocks: "))

for i in range(n):
    name = input("Enter stock name: ")
    quantity = int(input("Enter quantity: "))
    price = float(input("Enter price per stock: "))

    value = quantity * price
    total += value

    portfolio.append({
        "name": name,
        "quantity": quantity,
        "price": price,
        "value": value
    })

print("\n--- Stock Portfolio ---")

for stock in portfolio:
    print("Stock:", stock["name"])
    print("Quantity:", stock["quantity"])
    print("Price:", stock["price"])
    print("Total Value:", stock["value"])
    print()

print("Total Investment Value:", total)
```
